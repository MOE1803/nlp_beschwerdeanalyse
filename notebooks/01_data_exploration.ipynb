```python
print("Fehlende Werte pro Spalte:")
print(df.isnull().sum())

print("\nVerteilung der vordefinierten Kategorien (category_parent):")
print(df['category_parent'].value_counts())

print(f"\nDurchschnittliche Textlänge (Zeichen): {df['text_raw'].str.len().mean():.0f}")
```

```python
vor_dedup = len(df)
df = df.drop_duplicates(subset='text_raw').reset_index(drop=True)
print(f"Duplikate entfernt: {vor_dedup - len(df)}")

MIN_LAENGE = 20  # Mindestanzahl an Zeichen; kannst du anpassen und die Wahl im Bericht begründen
vor_filter = len(df)
df = df[df['text_raw'].str.len() >= MIN_LAENGE].reset_index(drop=True)
print(f"Kurztexte entfernt: {vor_filter - len(df)}")
print(f"Verbleibende Datensätze: {len(df)}")
```
